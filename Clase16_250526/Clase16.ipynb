**Integración**

Recordemos que la Integración en cálculo, es una generalización de la suma de infinitos sumandos, infinitesimalmente pequeños, en sí, una suma continúa.

El cálculo integral se utiliza principalmente para el cálculo de áreas y volúmenes de regiones y sólidos de revolución.

Dada una función f(x) de una variable real x y un intérvalo [a,b] de la recta real, la integral es igual al área de la región del plano xy limitada entre la gráfica de f, el eje x y las líneas verticales x=a y x=b; las áreas por debajo del eje x son negativas (wikipedia).
 Vease la figura

![](Integral1.png)

Matemáticamente la integración se representa por:

$\int_{a}^{b} f(x) = área\; bajo\; la\; curva$

La integral, también se define como la operación inversa a la derivada, por ejemplo:

La velocidad de una partícula que se mueve según la función y(t) es la derivada:

$v(t) = \frac{dy}{dt}$

Podemos también, podemos determinar la posición y(t) sabiendo la velocidad:

$y(t) = \int_{0}^{t}v(t)dt$

Esto es equivalente a resolver una ecuación diferencial:

$\frac{dy}{dt} = f(x)$

Con y(0) como condición inicial. 

(Referencia: Chapra)

**Métodos para integrar**

Para integrar una función, se dan las situaciones siguientes:

* Queremos integrar una función simple (trigonométrica, polinomial, etc)
* Queremos integrar una función complicada, dificil de integrar o diferenciar directamente
* Una función tabulada, es decir, tenemos los datos de x y f(x), pueden venir de datos experimentales.

Para el primer caso, existen métodos para "resolver" la integral analíticamente; pero para el segundo caso, hallar la "solución" analítica es impráctico o imposible; y para el tercer caso, se debe utilizar métodos aproximados.

**Métodos de Newton Cotes**

Los métodos de Newton Cotes, resuelven las integrales aproximando una función:

$I = \int_{a}^{b}f(x)dx = \int_{a}^{b}f_n(x)dx $

Donde $f_n(x)$ es un polinomio de la forma:

$f_n(x) = a_0 + a_1x + ... a_{n-1}x^{n-1} + a_{n}x^{n} $

**Método del trapecio**

Es la primera fórmula de integración de los métodos de Newton Cotes, el polinomio es de primer orden:

$I =  \int_{a}^{b}[f(a) + \frac{f(b) - f(a)}{b - a}(x-a)]dx $

Esto da como resultado:

$I = (b-a)\frac{f(a) + f(b)}{2}$ (esto es muy simple)

![](Integral3.png)

Si dividimos el intervalo de integración en varios segmentos y aplicamos la fórmula, siendo n el número de intervalos, tenemos:

![](Integral4.png)

# $ I = (b-a)\frac{f(a) + 2\sum^{n-1}_{i = 1}f(x_i) + f(x_n) }{2n}$

Donde:
$x_i = x+ih $

$ n = \frac{b-a}{h}$

y h es el paso.

```python
## Método del trapecio
# Practica 17

#integral con el metodo del trapecio

def Trapecio(f, a, b, N):
    sum = 0
    h = (b-a)/N
    for i in range(1,N):
        sum += f(a+ i*h)

    integr = h*(f(a)+f(b)+2*sum)/2
    print(integr)
    return integr

import math as mt
def f(x):
    return mt.sin(x)

Trapecio(f, 0 , 2, 1000)

```

**Regla de Simpson 1/3**

En este caso, utilizamos una función de segundo orden:

$ I = \int^{b}_{a} f(x)dx = \int^{b}_{a} f_2(x)dx$

Aplicando una vez, tenemos:

$I = (b-a)\frac{f(x_0) + 4f(x_1) + f(x_2)}{6}$

Aplicando múltiples veces dividiendo el intervalo en: $n = \frac{b-a}{h}$ subintervalos:

# $I = (b-a)\frac{f(x_0) + 4\sum^{n-1}_{i= 1,3,5}f(x+ih) + 2\sum^{n-2}_{j= 2,4,6}f(x+jh) + f(x_n)}{3n}$

![](Integral5.png)

```python
## Metodo de Simpson 1/3
#Definimos la funcion

def Simpson(f, n, a, b):
    h = (b-a)/n # aqui h es el paso

    sumpar=0
    sumimpar=0
    for i in range(1,n):
        x=a+i*h
        if i%2==0:
            sumpar+=f(x)
        else:
            sumimpar+=f(x)
    I = (b-a)*((f(a)+f(b)+4*sumimpar+2*sumpar)/(3*n))
    return I

def f(x):
    return mt.sin(x)

Simpson(f,1000, 0 , 2)
```

**Método de  Romberg**

El método de Romberg es una técnica de integración numérica que permite obtener un resultado de alta precisión combinando resultados de menor precisión. En lugar de calcular una sola integral complicada, calculamos varias integrales simples (usando la regla del trapecio) y las combinamos para "eliminar" el error.

Para aplicarla, calculamos la integral con N, 2N, 4N . ..  intervalos
Luego, mediante la extrapolación de Richardson, calculamos la integral nueva:

# $I_{j,k} = \frac{4^{k-1}I_{j+a,k-1} - I_{j,k-1}}{4^k - 1}$

Donde: 

$I_{j,k}$  es la nueva integral

$I_{j+1,k-1}$ es la estimación más precisa

$I_{j,k-1}$ es la estimación menos precisa

$k$ es el nivel de refinamiento (la columna)

$j$ es el indice de la fila

![](Integral6.png)

Para empezar, generamos la primera columna con el método de los trapecios, luego, utilizando el algoritmo vamos llenando la tabla:



```python
##Metodo de romberg

#Definimos la funcion del metodo de trapecios con un paso h:

def Trapecio(f, a, b, N):
    sum = 0
    h = (b-a)/N
    for i in range(1,N):
        sum += f(a+ i*h)

    integr = h*(f(a)+f(b)+2*sum)/2
    print(integr)
    return integr

#Ahora definimos la del metodo de romberg

def ROMBERG(f, a, b, n):
    R = [[0.0] * n for _ in range(n)]

    for i in range(n):
        h = (b-a)/(2**i)
        #Generamos la primera columna
        R[i][0] = Trapecio(f,a,b,2**i)
    #Ahora llenamos la tabla
    for k in range(1,n):
        for j in range(n-k):
            R[j][k] = (4**k * R[j+1][k-1] - R[j][k-1])/ (4**k -1)
    for X in R:
        print(X)

    return R[0][n-1]

def f(x):
    return mt.sin(x)

ROMBERG(f, 0 , 2, 10)

```

**Métodos de Runge Kutta**

Los métodos de Runge Kutta se utilizan para resolver EDOs de la forma:

# $\frac{dy}{dx} = f(x,y)$

Se aproxima la solución según:

$y_{i+1} = y_i + \phi h$

Gráficamente:

![](Edos1.png)

**Método de Euler**

La primera aproximación es haciendo:

$\phi = f(x_i, y_i)$

Así la solución queda:

$y_{i+1} = y_i + f(x_i,y_i)h$

Su error es:

$E = \frac{f'(x_i,y_i)}{2!} h^2 = 0(h^2)$


```python
#Metodo de Euler 


import numpy as np
#solucion practica 19
def df (x,y):
    return y/x + y*np.sin(x) + x*np.cos(x)*np.exp(-np.cos(x))
def freal(x):
    return x*np.sin(x)*np.exp(-np.cos(x))

# Definimos los valores iniciales
x0 = 0.5
y0 = 0.0996694
h = 0.2
real = freal(x0)
y_eu = y0
print("x, y_real, y_euler")
while x0 <= 0.9:
    print(f"{x0:.4f}, {real:.4}, {y_eu:.4f} ")
    #Calculamos el predictor (euler normal)
    y_eu = y0 + df(x0,y0)*h
    y0 = y_eu
    x0+=h
    real = freal(x0)
```

**Método de Heun**

Es una corrección al método de euler, primero realizamos una estimación:

$y_{i+1}^{0} = y_i + f(x_i, y_i)h$ Este se llama predictor

Y utilizando una especie de "promedio", calculamos el punto siguiente:

$y_{i+1}^{0} = y_i + \frac{f(x_i,y_i)+f(x_{i+1},y_{i+1}^{0})}{2}h$ Este se llama corrector

El criterio de convergencia suele ser la comparacion del valor

```python
#Metodo de heun


import numpy as np
#solucion practica 19
def df (x,y):
    return y/x + y*np.sin(x) + x*np.cos(x)*np.exp(-np.cos(x))
def freal(x):
    return x*np.sin(x)*np.exp(-np.cos(x))

# Definimos los valores iniciales
x0 = 0.5
y0 = 0.0996694
h = 0.1
real = freal(x0)
y_h = y0
print("x, y_real, y_heun")
while x0 <= 0.9:
    print(f"{x0:.4f}, {real:.4}, {y_h:.4f} ")
    #Calculamos el predictor (euler normal)
    y_h_0 = y0 + df(x0,y0)*h
    #Calculamos el corrector
    y_h = y0 + (df(x0,y0)+df(x0+h, y_h_0))*(h/2)
    y0 = y_h
    x0+=h
    real = freal(x0)
```

**Metodos de Runge Kutta**

La forma general de la solución es:

$y_{i+1}^{0} = y_i + \phi h$

Donde:

$\phi = a_{1}k_{1} + a_{2}k_{2} + ... + a_{n}k_{n} $

Donde las a's son constantes y las k's son:

$k1 = f(x_i, y_i)$

$k2 = f(x_i+p_1h, y_i+q_{11}k_1h)$

$k3 = f(x_i+p_2h, y_i + q_{21}k_1h + q_{22}k_2h)$

. . .

$kn = f(x_i + p_{n-1}h, y_i + q_{n-1,1}k_1h + q_{n-1,2}k2h + . . . + q_{n-1,n-1}k_{n-1}h )$

Donde las q's y p's son constantes. 

El índice de k's que se utilize, determia el orden del método.

**Métodos de Runge Kutta de Segundo orden**

La versión de segundo orden son:

$y_{i+1}^{0} = y_i + (a_1k_1 + a_2k_2) h$

Donde: 

$k1 = f(x_i, y_i)$

$k2 = f(x_i+p_1h, y_i+q_{11}k_1h)$

Donde las constantes se relacionan de la siguiente forma:

$a_1 + a_2 = 1$ 

$a_2 p_1= 1/2$

$a_2 q_{11}= 1/2$

Por ejemplo, el metodo de Heun es cuando $a_2 = 1/2$

El método de punto medio es cuando $a_2 = 1$

El método de Ralston es cuando $a_2 = 2/3$

```python
#Metodo de Heun 

import numpy as np
#solucion practica 19
def df (x,y):
    return y/x + y*np.sin(x) + x*np.cos(x)*np.exp(-np.cos(x))
def freal(x,y):
    return x*np.sin(x)*np.exp(-np.cos(x))

# Definimos los valores iniciales
x0 = 0.5
y0 = 0.0996694
h = 0.2

y0_c = y0
y_eu = y0
y_eu_c = y0
while x0 <= 0.9:
    print(f"{x0:.4f}, {y_eu:.4f}, {y_eu_c:.4f}")
    #Calculamos el predictor (euler normal)
    y_eu = y0 + h*df(x0,y0)
    y_p_c = y0 + h*df(x0,y0_c)
    #Calculamos el corrector
    k1 = df(x0, y0_c)
    x0+=h
    k2 = df(x0, y_p_c)
    y_eu_c = y0_c + (h/2)*(k1 + k2)
    y0 = y_eu
    y0_c = y_eu_c
```
