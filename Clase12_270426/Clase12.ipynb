**Métodos abiertos**

En los métodos cerrados la raíz se encontraba en un intervalo, repitiendo los métodos podíamos estimar el valor verdadero de la raíz. Estos métodos se dicen que son convergentes porque llegan al resultado verdadero a medida que avanza el cálculo.

En contraste, los métodos abiertos se basan en fórmulas que requieren un solo valor inicial de x (o 2). Estos métodos suelen diverger (no llegar al resultado) pero suelen ser más rápidos que los métodos cerrados.

**Iteración de Punto fijo**

Consiste en convertir nuestra ecuación f(x)=0, a una ecuación de la forma:

$x = g(x)$ (en otras palabras, despejar una x)

Así, podemos estimar la raíz mediante la fórmula iterativa:

$x_{i+1} = g(x_i)$

Como esta es una fórmula iterativa, tenemos que calcular el error aproximado:

$e_a = |\frac{x_{i+1} - x_i}{x_{i+1}}|*100%$

**Ejemplo**

Dada la función $f(x) = e^{-x} - x$, despejamos x de $f(x) = 0$:

$x = e^{-x}$

Iterando tenemos:

```python
##Punto fijo
#Definimos nuestra funcion g(x):
import numpy as np
def ge(x):
    return np.exp(-x)

Graph(ge, -5, 5,10)

#Iteramos:
x_in = 0
xr = 10
err = abs((xr-x_in)/xr)
tol = 0.00001
it = 0
while err > tol:
    xr = ge(x_in)
    err = abs((xr-x_in)/xr)
    x_in = xr
    print(xr,err,it, sep="   ")
    it +=1
print("Finales")
print(xr,err,it, sep="   ")

```

**Newton Raphson**

Este método se deriva de una interpretación geométrica de la derivada, de la definición de la derivada entre un intérvalo $x_i$ y $x_{i+1}$:

$f'(x_i) = \frac{f(x_{i}) - f(x_{i+1})}{x_{i} - x_{i+1}}$

Si la raíz se encuentra en $x_{i+1}$, operando tenemos:

$f'(x_i) = \frac{f(x_{i}) - 0}{x_{i} - x_{i+1}}$

# $x_{i+1} = x_i - \frac{f(x_{i})}{f'(x_{i+1})}$

### Esta es la fórmula de Newton Raphson

```python
#Método de Newton Raphson

import numpy as np

#Definimos una función:

def Ge(x):
    return np.exp(-x) -x

#Y su derivada

def d_Ge(x):
    return -np.exp(-x)-1

# Graph(ge, -5, 5,10)
# Graph(d_ge, -5, 5,10)

#Iteramos:
x_in = 0
xr = 10
err = abs((xr-x_in)/xr)
tol = 0.00001
it = 0
while err > tol:
    xr = x_in - Ge(x_in)/d_Ge(x_in)
    err = abs((xr-x_in)/xr)
    x_in = xr
    print(xr,err,it, sep="   ")
    it +=1
    if it > 100000:
        break
print("Finales")
print(xr,err,it, sep="   ")

```

**Método de la Secante**

Una ventaja del método de Newton Raphson es el cálculo de la derivada, sin embargo, esto puede no funcionar para polinomios o ciertas funciones.

Para estas funciones, cuya derivada es difícil de evaluar, se aproxima como una división de diferencias:

$f'(x_i) = \frac{f(x_{i-1}) - f(x_{i})}{x_{i-1} - x_{i}}$

Reemplazando en la ecuación de Newton Raphson:

# $x_{i+1} = x_i - \frac{f(x_{i})*(x_{i-1}-x_{i})}{f(x_{i-1}) - f(x_{i})}$

### Esta es la fórmula del método de la secante
