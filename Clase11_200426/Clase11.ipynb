**Raíces de Ecuaciones**

Dada una función f(x), la raíz de la "ecuación" es el valor de x que hace que f(x) = 0.

Por ejemplo la ecuación cuadrática:

$f(x) = ax^2 + bx + c$

Sus raíces son:

# $x_{1,2} = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$

Ya que: $ax^2_{1,2} + bx_{1,2} + c = 0$

La ecuación cuadrática es un caso particular, sin embargo, existen muchas ecuaciones cuyas raíces no son fáciles de calcular.

**Método de Bisección**

El principio fundamental del método de bisección y falsa posición, es: 

Dada una función f(x) y dos puntos, a y b. Sí:

$f(a)*f(b) < 0$

Entonces, existe una raiz en el intervalo [a,b]

En el método de bisección el intervalo se va dividiendo a la mitad, es decir:

# $c = \frac{a+b}{2}$

Entonces, el algoritmo es:

* Elegir un intervalo [a,b] tal que $f(a)*f(b) < 0$.
* Calcular $c = \frac{a+b}{2}$.
* Evaluamos: $f(c)*f(b)$ y $f(a)*f(c)$, cambiamos el intervalo al que resulte menor a 0.
* Repetimos


![](Bisection_anime.gif)

```python
###Metodo de biseccion
###Daniel ROberto Garcia Miranda
###7363560-1I

import math #Importamos para usar el exponencial

#Definimos la funcion
def f(x):
    xd=-x+math.exp(-x)
    return xd

#Colocamos nuestros parametros iniciales
VERDADERO=0.567143290409783
xa = 0 #Punto inicial
xb = 1 #Punto final
xr = (xa+xb)/2 #Punto medio
tol = 0.000000000000001 #Tolerancia o error
xdd=1 #Parametro auxiliar
err = abs((VERDADERO - xr)/VERDADERO) #error absoluto/verdadero, usar si tenemos la solución exacta
erra = abs((xr - xdd)/xr) #error aproximado, usar sí no tenemos la solución exacta
numero = 0

#en la vida real usamos el error aproximado, ya que podemos no tener el valor verdadero

#Creamos el bucle, hasta que el error sea menor a la tolerancia
while err>tol:  #Condicion de que el error verdadero tenga al menos 15 decimales
    xr = (xa+xb)/2
    if f(xr)*f(xa)<0:
        xb=xr
    else: 
        xa=xr
    err = abs((VERDADERO - xr)/VERDADERO)
    erra = abs((xr - xdd)/xr)
    xdd=xr
    print(numero, xr, err, erra)
    numero +=1
    # print('xa = ',xa)
    # print('xb = ',xb)
    # print('xr = ',xr)
    # print("error = ", err)
    # print("xr = ",xr)  

#Imprime el resultado
print("xr = ",xr)
print("Error Verdadero = ",err)
print("Error Aproximado= ",erra)
print("Número de iteraciones ",numero)
```

Ahora nuestro salto será de:


# $c = b - \frac{f(b)*(b - a)}{f(b) - f(a)}$
