# Sistemas de ecuaciones lineales

Un sistema lineal puede escribirse como:

$$
A\mathbf{x} = \mathbf{b}
$$

Donde:

- $A$: matriz de coeficientes  
- $\mathbf{x}$ : vector incógnita  
- $\mathbf{b}$: vector independiente  

---

**Método de Gauss - Jordan**

**Idea**

Transformar la matriz en una matriz triangular superior mediante eliminación.

---

**Pasos**

1. Elegir pivote  
2. Hacer ceros debajo del pivote  
3. Continuar fila por fila  
4. Resolver por sustitución hacia atrás  

**Método de Gauss-Seidel**

**Idea**

Resolver el sistema iterativamente:

$$
x_i^{(k+1)} =
\frac{1}{a_{ii}}
\left(
b_i -
\sum_{j \ne i} a_{ij}x_j
\right)
$$

---

**Características**

- Método iterativo  
- Usa inmediatamente los valores nuevos  
- Requiere aproximación inicial  

---

**Condición importante**

La matriz debe ser diagonal dominante:

$$
|a_{ii}| > \sum_{j \ne i}|a_{ij}|
$$

**Inversión de matrices** 
Queremos encontrar una matriz:

$$
A^{-1}
$$

tal que:

$$
AA^{-1} = I
$$

donde \( I \) es la matriz identidad.

---

```python
##Método de Gauss Jordan
#Daniel Roberto Garcia Miranda

def gauss_jordan(M):
    n = len(M)  ## numero de filas, ecuaciones
    m = len(M[0])  ## numero de columnas, variables + termino independiente
    for k in range(n):
        pivote = M[k][k]
        for j in range(m):
            M[k][j] /= pivote ##Normalizamos el pivote
        for i in range(n):
            if i != k: 
                factor = M[i][k]
                for j in range(m):
                    M[i][j] -= factor * M[k][j] #Restamos a las demás lineas
    XD = [pacman[-1] for pacman in M] # extraemos el vector solución
    return XD
M = [[2, 1, -1, 8],
     [-3, -1, 2, -11],
     [-2, 1, 2, -3]]
gauss_jordan(M)
```

**Métodos de búsqueda de raíces**

Queremos resolver:

$$
f(x)=0
$$

---

**Métodos cerrados**

**Idea**

Trabajan en un intervalo:

$$
[a,b]
$$

donde:

$$
f(a)f(b)<0
$$

Existe al menos una raíz en el intervalo.

---

**Método de Bisección**

**Idea**

Dividir repetidamente el intervalo a la mitad.

---

**Fórmula**

$$
x_r = \frac{a+b}{2}
$$

---

**Pasos**

1. Calcular punto medio  
2. Evaluar signo  
3. Elegir nuevo intervalo  
4. Repetir  

---

**Ventajas**

- Siempre converge  
- Muy estable  

---

**Desventajas**

- Convergencia lenta  

---

**Método de Falsa Posición (Interpolación Lineal)**

**Idea**

Aproximar la función con una recta.



**Fórmula**

$$
x_r =
b -
\frac{f(b)(a-b)}
{f(a)-f(b)}
$$



**Ventajas**

- Más rápido que bisección  
- Mantiene el intervalo cerrado  

---

**Desventajas**

- Puede estancarse en algunos casos  

---

```python
###Metodo de biseccion
###Daniel ROberto Garcia Miranda

import math

#Definimos la funcion
def f(x):
    xd=-x+math.exp(-x)
    return xd

def Biseccion(f, a, b, tol = 0.0000001):
    err = 1
    xdd = 1 #parametro auxiliar
    it = 0
    while err > tol:
        xr = (a+b)/2
        if f(xr)*f(a)<0:
            b = xr
        else: a = xr
        err = abs((xr - xdd)/xr)
        xdd = xr
        it +=1
        print(it, xr, err)
    return xr

Biseccion(f, 0, 1,)
```

**Método de Newton-Raphson**

**Fórmula**
$$
x_{n+1} = x_n - \frac{2*f(x_n)*f'(x_n)}{2*[f'(x_n)]^2 - f(x_n)*f''(x_n)}
$$ 

---

**Interpretación**

La raíz se aproxima usando la recta tangente.

---

**Ventajas**

- Muy rápido  
- Convergencia cuadrática  

---

**Desventajas**

- Requiere derivada  
- Sensible a la semilla inicial  
- Puede divergir  


```python
#Funcion de Newton Raphson
#Tendremos como parametros la funcion, su derivada, la semilla y su tolerancia
def NR(f, df, xin = 0, tol = 0.000001, impr = True):
    print("---- Método de Newton Raphson ----")

    err = 1
    it = 0
    while err > tol:
        xr = xin - f(xin)/df(xin)
        err = abs((xr-xin)/xr)
        xin = xr
        if impr:
            print(xr,err,it, sep="   ")
        it +=1
        if it > 100000:
            break
    print("La raíz es: ")
    print(xr, err, it, sep = "  ")

#Funcion del metodo de la secante modificado
#Tendremos como parametros la funcion, la semilla, el paso y su tolerancia
def SECm(f, xin = 0, h = 0.1, tol = 0.000001, impr = True):
    print("---- Método de la Secante modificada ----")
    err = 1
    it = 0
    while err > tol:
        xr = xin - (f(xin)*(h))/(f(xin+h)- f(xin))
        err = abs((xr-xin)/xr)
        xin = xr
        if impr:
            print(xr,err,it, sep="   ")
        it +=1
        if it > 100000:
            break
    print("La raíz es: ")
    print(xr,err,it, sep="   ")

```
