**Newton Raphson**

Este método se deriva de una interpretación geométrica de la derivada, de la definición de la derivada entre un intérvalo $x_i$ y $x_{i+1}$:

$f'(x_i) = \frac{f(x_{i}) - f(x_{i+1})}{x_{i} - x_{i+1}}$

Si la raíz se encuentra en $x_{i+1}$, operando tenemos:

$f'(x_i) = \frac{f(x_{i}) - 0}{x_{i} - x_{i+1}}$

# $x_{i+1} = x_i - \frac{f(x_{i})}{f'(x_{i+1})}$

### Esta es la fórmula de Newton Raphson
**Método de la secante**

Una ventaja del método de Newton Raphson es el cálculo de la derivada, sin embargo, esto puede no funcionar para polinomios o ciertas funciones.

Para estas funciones, cuya derivada es difícil de evaluar, se aproxima como una división de diferencias:

$f'(x_i) = \frac{f(x_{i-1}) - f(x_{i})}{x_{i-1} - x_{i}}$

Reemplazando en la ecuación de Newton Raphson:

# $x_{i+1} = x_i - \frac{f(x_{i})*(x_{i-1}-x_{i})}{f(x_{i-1}) - f(x_{i})}$

### Esta es la fórmula del método de la secante

Si tomamos un paso $h = x_{i} - x_{i-1}$, la derivada sería: 

$f'(x_{i}) = \frac{f(x_{i} + h) - f(x_{i})}{h}$

Así:

# $x_{i+1} = x_i - \frac{f(x_{i})*h}{f(x_{i}+h) - f(x_{i})}$

```python
#Funcion de Newton Raphson
#Tendremos como parametros la funcion, su derivada, la semilla y su tolerancia
def NR(f, df, xin = 0, tol = 0.000001, impr = True):
    print("---- Método de Newton Raphson ----")

    err = 1
    it = 0
    while err > tol:
        xr = xin - f(xin)/df(xin)
        err = abs((xr-xin)/xr)
        xin = xr
        if impr:
            print(xr,err,it, sep="   ")
        it +=1
        if it > 100000:
            break
    print("La raíz es: ")
    print(xr, err, it, sep = "  ")

#Funcion del metodo de la secante modificado
#Tendremos como parametros la funcion, la semilla, el paso y su tolerancia
def SECm(f, xin = 0, h = 0.1, tol = 0.000001, impr = True):
    print("---- Método de la Secante modificada ----")
    err = 1
    it = 0
    while err > tol:
        xr = xin - (f(xin)*(h))/(f(xin+h)- f(xin))
        err = abs((xr-xin)/xr)
        xin = xr
        if impr:
            print(xr,err,it, sep="   ")
        it +=1
        if it > 100000:
            break
    print("La raíz es: ")
    print(xr,err,it, sep="   ")

#Función Graficadora
import numpy as np
import matplotlib.pyplot as plt
#Funcion a graficar
def g(x):
    xd = x**10 - 1
    return xd
#Funcion
def Graph(FUNCION,min, max, N=10, Titulo = "Grafico"):
    X = np.linspace(min,max,N)
    Y = FUNCION(X)
    plt.figure(figsize=(6,6))
    plt.plot(X,Y)
    plt.title(Titulo)
    plt.grid()
    plt.xlabel("x")
    plt.ylabel("y")
    plt.show()

```

Podemos observar que los métodos de **Newton-Raphson (NR)** y **Secante** presentan dificultades en ciertas situaciones:

- **Raíces múltiples:**  
  La convergencia se vuelve más lenta y puede perder su orden teórico (por ejemplo, Newton deja de ser cuadrático).

- **Alta dependencia de la semilla inicial:**  
  Una mala elección del valor inicial puede llevar a convergencia hacia otra raíz o incluso a divergencia.

- **Convergencia lenta o ineficiente:**  
  En funciones mal condicionadas o con comportamiento plano cerca de la raíz, el método puede avanzar muy lentamente.

- **Problemas con la derivada:**  
  - Si $f'(x) \approx 0 )$: se producen pasos muy grandes e inestabilidad.  
  - Si $f'(x) \to \infty)$: el método puede volverse errático o no converger.

- **Comportamiento oscilatorio:**  
  En algunos casos, las iteraciones pueden entrar en ciclos y no acercarse a la raíz.

---

> Aunque estos métodos son muy eficientes, su desempeño depende fuertemente de la función y de la elección inicial, por lo que siempre es necesario analizar la convergencia antes de aplicarlos.

**Newton Raphson de segundo orden**

$x_{n+1} = x_n - \frac{2f(x_n)f'(x_n)}{2[f'(x_n)]^2 - f(x_n)f''(x_n)}$

```python
#Newton Raphson de segundo orden
#Funcion de Newton Raphson
#Tendremos como parametros la funcion, su derivada, su segunda derivada, la semilla y su tolerancia
def NR2O(f, df, ddf, xin = 0, tol = 0.000001, impr = True):
    print("---- Método de Newton Raphson de segundo orden ----")

    err = 1
    it = 0
    while err > tol:
        xr = xin - (2*f(xin)*df(xin))/(2*df(xin)**2 - f(xin)*ddf(xin))
        err = abs((xr-xin)/xr)
        xin = xr
        if impr:
            print(xr,err,it, sep="   ")
        it +=1
        if it > 100000:
            break
    print("La raíz es: ")
    print(xr, err, it, sep = "  ")

#Para la funcion
def f(x):
    return np.exp(0.5*x) - 3*np.cos(2*x)
def df(x):
    return 0.5*np.exp(0.5*x) + 6*np.sin(2*x)
def ddf(x):
    return 0.25*np.exp(0.5*x) + 12*np.cos(2*x)

Graph(f,-3,3,155)
NR2O(f,df,ddf,-3)
NR2O(f,df,ddf,-1)
NR2O(f,df,ddf,0)
```

**Derivadas**

Podemos escribir la derivada de 3 formas:

Derivada hacia adelante: 

# $f'(x) \approx \frac{f(x+h) - f(x)}{h}$

Derivada hacia atrás:

# $f'(x) \approx \frac{f(x) - f(x-h)}{h}$

Derivada intermedia o centrada:

# $f'(x) \approx \frac{f(x+h) - f(x-h)}{2*h}$

El error va a depender del parámetro h:

* Sí h es grande, habrá error de aproximación
* Sí h es pequeño, habrá error de redondeo
