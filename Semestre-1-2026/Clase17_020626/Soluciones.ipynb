```python
import numpy as np
#solucion practica 20 parte uno
def df (x,y):
    return y/x + y*np.sin(x) + x*np.cos(x)*np.exp(-np.cos(x))
def freal(x):
    return x*np.sin(x)*np.exp(-np.cos(x))

# Definimos los valores iniciales
x0 = 0.5
y0 = 0.0996694
h = 0.1

y0_c = y0
y_eu = y0
y_eu_c = y0
while x0 <= 20:
    print(f"{x0:.4f}, {freal(x0):.4f}, {y_eu:.4f}, {y_eu_c:.4f}")
    #Calculamos el predictor (euler normal)
    y_eu = y0 + h*df(x0,y0)
    y_p_c = y0_c + h*df(x0,y0_c)
    #Calculamos el corrector
    k1 = df(x0, y0_c)
    x0+=h
    k2 = df(x0, y_p_c)
    y_eu_c = y0_c + (h/2)*(k1 + k2)
    y0 = y_eu
    y0_c = y_eu_c
```
