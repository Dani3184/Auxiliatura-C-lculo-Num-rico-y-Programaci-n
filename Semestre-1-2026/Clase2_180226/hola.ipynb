**Uso de funciones en python**

Una función es un subprograma que realiza una tarea específica
Ya existen funciones predefinidas como:
- `len()` te da la longitud de un string
- type() te dice el tipo del objeto
- print() escribe en la terminal


Una función puede”recibir” un parámetro y producir un resultado (retornar un valor).
Por ejemplo len() recibe un string como parámetro y te da como resultado (te retorna) un numero tipo int.


**Métodos en python**

Un método es una función que se aplica a un objeto específico:
- first_name.upper()
- first_name.count(“a”)
- first_name.replace(“a”, “e”)


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

def Elipse(h,k,a,b,x,y):
    #Calculamos:
    dx = (b**2)*(x-h)**2
    dy = (a**2)*(y-k)**2
    if dx + dy <= (a**2)*(b**2):
        return True
    else: return False

#definimos los parametros de la elipse
a = 2
b = 3
h = 1
k = 5
#el punto
x = int(input("Introduzca x: "))
y = int(input("Introduzca y: "))

if Elipse(a,b,h,k,x,y): print("El punto esta dentro la elipse")
else: print("El punto esta fuera de la elipse")

#Grafico
# 1. Dibujar la elipse (forma ultra simple)
t = np.linspace(0, 7, 100) 
plt.plot(h + a*np.cos(t), k + b*np.sin(t))

# 2. Dibujar el punto
color = 'green' if Elipse(h,k,a,b,x,y) else 'red'
plt.scatter(x, y, c=color)

# 3. Mostrar
plt.xlabel("x")
plt.ylabel("y")
plt.axis('equal')
plt.show()
```
