```python
def gauss_jordan(M):
    n = len(M)  ## numero de filas, ecuaciones
    m = len(M[0])  ## numero de columnas, variables + termino independiente
    for k in range(n):
        pivote = M[k][k]
        for j in range(m):
            M[k][j] /= pivote ##Normalizamos el pivote
        for i in range(n):
            if i != k: 
                factor = M[i][k]
                for j in range(m):
                    M[i][j] -= factor * M[k][j] #Restamos a las demás lineas

    XD = [pacman[-1] for pacman in M] # extraemos el vector solución
    return XD
M = [[2, 1, -1, 8],
     [-3, -1, 2, -11],
     [-2, 1, 2, -3]]
gauss_jordan(M)
```

```python
#Funcion de Newton Raphson
#Tendremos como parametros la funcion, su derivada, la semilla y su tolerancia
def NR(f, df, xin = 0, tol = 0.000001, impr = True):
    print("---- Método de Newton Raphson ----")

    err = 1
    it = 0
    while err > tol:
        xr = xin - f(xin)/df(xin)
        err = abs((xr-xin)/xr)
        xin = xr
        if impr:
            print(xr,err,it, sep="   ")
        it +=1
        if it > 100000:
            break
    print("La raíz es: ")
    print(xr, err, it, sep = "  ")

#Funcion del metodo de la secante modificado
#Tendremos como parametros la funcion, la semilla, el paso y su tolerancia
def SECm(f, xin = 0, h = 0.1, tol = 0.000001, impr = True):
    print("---- Método de la Secante modificada ----")
    err = 1
    it = 0
    while err > tol:
        xr = xin - (f(xin)*(h))/(f(xin+h)- f(xin))
        err = abs((xr-xin)/xr)
        xin = xr
        if impr:
            print(xr,err,it, sep="   ")
        it +=1
        if it > 100000:
            break
    print("La raíz es: ")
    print(xr,err,it, sep="   ")

#Función Graficadora
import numpy as np
import matplotlib.pyplot as plt
#Funcion a graficar
def g(x):
    xd = x**10 - 1
    return xd
#Funcion
def Graph(FUNCION,min, max, N=10, Titulo = "Grafico"):
    X = np.linspace(min,max,N)
    Y = FUNCION(X)
    plt.figure(figsize=(6,6))
    plt.plot(X,Y)
    plt.title(Titulo)
    plt.grid()
    plt.xlabel("x")
    plt.ylabel("y")
    plt.show()

```
