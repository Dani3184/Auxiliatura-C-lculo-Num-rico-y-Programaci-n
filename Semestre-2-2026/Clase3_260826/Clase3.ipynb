**Vectores y arreglos**

En programación podemos llamar a los "arreglos" vectores.
No los llamamos directamente vectores, ya que no son necesariamente un conjunto de "números" que representan una magnitud vectorial (como en física). Sino que es un arreglo de algún tipo de variable (int, float, etc).

En python existen arreglos con diferentes tipos en cada posición (es decir, en la posición 0 puede haber un entero, y en la posición 1 una palabra: Ej. [1, "hola"]). Para nuestros fines, (espero) no utilicemos este tipo de vectores

El código siguiente muestra varios ejemplos de las diferentes cosas que podemos hacer con un vector:

```python
#Vectores y arreglos
# Creamos dos arreglos (arrays)

A = [1 , 2, 3, 4, 5] 
B = [6 , 7, 8, 9, 10] 
# Imprimimos una componente:
print(A[0])
# Imprimimos la ultima y penultima componente:
print(A[-1])
print(A[-2])
# Imprimimos un pedazo de todas las componentes
print(A[1:3]) # De la 2da a la 3ra
print(A[:3]) # De la 1ra a la 3ra

# Imprimimos la suma de 2 componentes:
print(A[0]+B[0])
# Juntamos los vectores
print(A+B)
# Podemos crear un vector vacio
V = []
# Y añadir componentes al final
V.append(1)
print(V)
V.append(2)
print(V)
V.append("hola")
print(V)
# O en una posición específica
V.insert(1,2) # añade un 2 en la posición 1
print(V)
print(f"La longitud del vector A es: {len(A)}")
```

**Lectura y escritura de archivos**

Para leer y escribir archivos, tenemos que utilizar la función with y open:

```python
#Ejemplo de escritura de archivos

with open("Archivo.txt", "w") as Archivo:
    for i in range(10):
        x = i**2
        Archivo.write(f"{i} {x}\n")
```

```python
#Ejemplo de lectura de archivos
# Ahora ponemos "r" de read en la función open

x = []
y = []

with open("Archivo.txt", "r") as Archivo:
    print(Archivo)
    for linea in Archivo:
        valores = linea.split()
        print(valores)
        x.append(float(valores[0]))
        y.append(float(valores[1]))
print(x)
print(y)

```
