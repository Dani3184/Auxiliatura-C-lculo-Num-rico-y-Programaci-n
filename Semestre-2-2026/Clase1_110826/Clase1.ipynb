**Condicional**

La funcion if nos permite realizar procesos condicionados a un valor booleano, que puede resultar de una comparación (igual que, menor que, mayor que, diferente que) u operaciones logica (and, or). 

Si la condicion que se escribe a lado de la funcion if es verdadera, se ejecuta el codigo que le sigue, si no, se debe llamar a la funcion "else", y escribir el codigo.

Se puede escribir una cadena de condicionales utilizando la funcion "elif":



```python
## Condicionales
# Funcion if

Numero = 5
if Numero > 3:
    print("XD")
else: 
    print("X,D")

# Podemos hacer mas condiciones, ahora con operaciones logicas
edad = int(input())
carrera = "fisica"

if edad > 18 and carrera == "fisica":
    print("Si sabes")
elif edad < 18 and carrera == "fisica":
    print("No sabes")
else:
    print("Sabes")
```

**Bucles**

Existen dos funciones con las que se pueden realizar bucles: 

* Funcion for: ejecuta el codigo dentro de la funcion hasta que la condicion se cumple, dentro de su condicion, se puede anadir procesos
* Funcion while: El bucle esta activo siempre que su condicion sea verdadera



```python
## Bucles
#Uso de for
print("Uso de for")
for i in range(3):
    print(i)
# Si tenemos un vector:
xd = [1, 2, 3, 4, 5]
for _ in xd: print(_)

#Uso de while
print("Uso de while")
x = 0
while(x <= 5):
    print(x)
    x+=1
```
