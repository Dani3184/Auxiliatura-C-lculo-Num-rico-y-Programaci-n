```python
# Solución ejercicio 1
# Un código que determine que un número es divisible por 4, sería muy simple utilizando solamente el operador %. 
# El criterio de divisibilidad por 4 (para grandes números) se basa en verificar que sus últimos 2 dígitos sean un múltiplo de 4
# En código quedaría como:

def div4(Numero):
    # obtenemos los ultimos 2 digitos con str e int
    dig2 = int(str(Numero)[-2:])
    if (dig2%4)!= 0:
        return False
    else: return True

print(div4(2414))
print(div4(2414642635))
print(div4(24141353613517538372312))
```

```python
# Ejercicio 2
# Es el típico ejercicio de la ecuación cuadrática, basado en el uso del discribimante D = B^2 - 4AC. 
# En este ejercicio, utilizaremos match:
import math as mt

def Cuadratica(A, B, C):
    Disc = B**2 - 4*A*C
    match Disc:
        case d if d > 0:
            x1 = (-B + mt.sqrt(Disc)) / (2*A)
            x2 = (-B - mt.sqrt(Disc)) / (2*A)
            print(f"La ecuacion tiene 2 soluciones reales: {x1} y {x2}")
        case 0:
            x = -B / (2*A)
            print(f"La ecuacion tiene 1 solucion real: {x}")
        case d if d < 0:
            real = -B/(2*A)
            img = mt.sqrt(-Disc)/(2*A)
            print(f"La ecuacion tiene 2 soluciones imaginarias: {real} + {img}i y {real} - {img}")

Cuadratica(1,1,1)
Cuadratica(1,2,5)
Cuadratica(1,1,0)


```
